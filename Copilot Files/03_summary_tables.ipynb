# Cross-Dataset Summary Tables (Copilot Files)

Loads generated result CSVs and creates a compact publication-ready table for all baselines and fusion schemes.

```python
import os
import pandas as pd

root = '/home/ritvik/Python/Copilot Files/results'
all_path = os.path.join(root, 'all_datasets_metrics_summary.csv')
df = pd.read_csv(all_path)
df.head()
```

```python
metric_cols = [
    'Accuracy', 'F1', 'AUPRC', 'ROC_AUC', 'TPR@0.1%FPR', 'TPR@1%FPR'
]
existing = [c for c in metric_cols if c in df.columns]
summary = (
    df[['Dataset', 'Model'] + existing]
    .sort_values(['Dataset', 'AUPRC'], ascending=[True, False], na_position='last')
    .reset_index(drop=True)
)
summary
```

```python
csv_out = os.path.join(root, 'final_publication_summary_table.csv')
summary.to_csv(csv_out, index=False)
print('Saved:', csv_out)
```
