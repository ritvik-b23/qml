```python
import matplotlib.pyplot as plt
import numpy as np

# Data points extracted from the image
epochs = [100, 150, 200, 250, 300, 350, 400, 450, 500, 550, 600, 650, 700, 750, 800]

# Training values (blue line)
training = [0.9383, 0.9365, 0.9410, 0.9345, 0.9290, 0.9360, 0.9350, 0.9345, 0.9310, 0.9315, 0.9320, 0.9328, 0.9330, 0.9330, 0.9340]

# Validation values (red line)
validation = [0.9220, 0.9342, 0.9342, 0.9343, 0.9343, 0.9345, 0.9345, 0.9343, 0.9343, 0.9343, 0.9343, 0.9342, 0.9343, 0.9343, 0.9343]

# Create the plot
fig, ax = plt.subplots(figsize=(10, 6))

# Plot lines
ax.plot(epochs, training, 'o-', color='blue', label='Training', linewidth=2, markersize=6)
ax.plot(epochs, validation, 'o-', color='red', label='Validation', linewidth=2, markersize=6)

# Add shaded confidence intervals
training_std = 0.008  # Approximate standard deviation for shading
validation_std = 0.008

ax.fill_between(epochs, 
                np.array(training) - training_std, 
                np.array(training) + training_std, 
                color='blue', alpha=0.15)

ax.fill_between(epochs, 
                np.array(validation) - validation_std, 
                np.array(validation) + validation_std, 
                color='red', alpha=0.15)

# Set axis limits
ax.set_xlim(50, 850)
ax.set_ylim(0.920, 0.950)

# Add grid
ax.grid(True, alpha=0.3)

# Add legend
ax.legend(loc='center right', frameon=True)

# Set labels
ax.set_xlabel('Epoch', fontsize=12)
ax.set_ylabel('Accuracy', fontsize=12)

# Set tick label font sizes
ax.tick_params(axis='both', which='major', labelsize=12)

# Set legend font size
ax.legend(loc='center right', frameon=True, fontsize=12)

# Adjust layout
plt.tight_layout()
plt.show()
```
